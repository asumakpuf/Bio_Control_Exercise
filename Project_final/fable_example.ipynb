### Connect RealSense camera
This cell uses Open3D's RealSense support to find the connected Intel RealSense camera, open it, capture one aligned RGB-D frame, and then stop the capture.

```python
from pathlib import Path
import open3d as o3d

# Open3D RealSense config. Leave serial empty to use the first connected camera.
realsense_config = {
    "serial": "",
    "color_format": "RS2_FORMAT_RGB8",
    "color_resolution": "640,480",
    "depth_format": "RS2_FORMAT_Z16",
    "depth_resolution": "640,480",
    "fps": "15",
    "visual_preset": "",
}

print("Connected RealSense devices:")
o3d.t.io.RealSenseSensor.list_devices()

rs_cfg = o3d.t.io.RealSenseSensorConfig(realsense_config)
rs = o3d.t.io.RealSenseSensor()
rs.init_sensor(rs_cfg, 0, "")  # sensor index 0, no .bag recording file
rs.start_capture(False)

try:
    im_rgbd = rs.capture_frame(True, True)  # wait for frames, align depth to color

    debug_dir = Path("debug_realsense")
    debug_dir.mkdir(exist_ok=True)
    o3d.t.io.write_image(str(debug_dir / "color.png"), im_rgbd.color)
    o3d.t.io.write_image(str(debug_dir / "depth_raw.png"), im_rgbd.depth)

    print("Captured RealSense RGB-D frame")
    print("Saved debug images to:", debug_dir.resolve())
    print("Color image:", im_rgbd.color)
    print("Depth image:", im_rgbd.depth)
finally:
    rs.stop_capture()
```

### Live RealSense preview and camera tuning
Run the next cell if `pyrealsense2` or `ipywidgets` is missing in the notebook kernel. Then run the live-preview cell below it. The preview starts immediately and the camera parameters update while the stream is running. Use **Stop RealSense live view** before rerunning other camera cells.

```python
# Uncomment and run once if the live-preview cell says a dependency is missing.
# %pip install pyrealsense2 ipywidgets
```

```python
import threading
import time

import cv2
import numpy as np
from IPython.display import display

try:
    import ipywidgets as widgets
except ImportError as exc:
    raise ImportError("Install ipywidgets first: %pip install ipywidgets") from exc

try:
    import pyrealsense2 as rs2
except ImportError as exc:
    raise ImportError("Install the Intel RealSense Python SDK first: %pip install pyrealsense2") from exc

if "realsense_config" not in globals():
    realsense_config = {
        "serial": "",
        "color_format": "RS2_FORMAT_RGB8",
        "color_resolution": "640,480",
        "depth_format": "RS2_FORMAT_Z16",
        "depth_resolution": "640,480",
        "fps": "15",
        "visual_preset": "",
    }

# Stop an older live stream if this cell is rerun.
if "_rs_live" in globals() and _rs_live.get("running"):
    _rs_live["running"] = False
    try:
        _rs_live["pipeline"].stop()
    except Exception:
        pass


def _parse_resolution(value, fallback):
    try:
        width, height = str(value).split(",")
        return int(width), int(height)
    except Exception:
        return fallback


def _sensor_named(device, name_part):
    name_part = name_part.lower()
    for sensor in device.query_sensors():
        try:
            sensor_name = sensor.get_info(rs2.camera_info.name).lower()
        except Exception:
            continue
        if name_part in sensor_name:
            return sensor
    return None


def _first_sensor_supporting(device, option):
    for sensor in device.query_sensors():
        if sensor.supports(option):
            return sensor
    return None


def _get_option(sensor, option, fallback=0.0):
    try:
        return sensor.get_option(option)
    except Exception:
        return fallback


def _make_option_slider(sensor, option, label, integer=False):
    if sensor is None or not sensor.supports(option):
        return None

    opt_range = sensor.get_option_range(option)
    default_value = getattr(opt_range, "default", getattr(opt_range, "def", opt_range.min))
    value = _get_option(sensor, option, default_value)
    step = opt_range.step if opt_range.step else 1

    slider_cls = widgets.IntSlider if integer else widgets.FloatSlider
    slider = slider_cls(
        value=int(value) if integer else float(value),
        min=int(opt_range.min) if integer else float(opt_range.min),
        max=int(opt_range.max) if integer else float(opt_range.max),
        step=max(1, int(step)) if integer else float(step),
        description=label,
        continuous_update=False,
        readout=True,
        style={"description_width": "120px"},
        layout=widgets.Layout(width="420px"),
    )

    def _on_change(change):
        if change["name"] == "value":
            try:
                sensor.set_option(option, float(change["new"]))
                status.value = f"<span>Set {label} to {change['new']}</span>"
            except Exception as exc:
                status.value = f"<span style='color:#b00020'>Could not set {label}: {exc}</span>"

    slider.observe(_on_change, names="value")
    return slider


def _make_option_checkbox(sensor, option, label):
    if sensor is None or not sensor.supports(option):
        return None

    checkbox = widgets.Checkbox(
        value=bool(_get_option(sensor, option, 1.0)),
        description=label,
        indent=False,
        layout=widgets.Layout(width="240px"),
    )

    def _on_change(change):
        if change["name"] == "value":
            try:
                sensor.set_option(option, 1.0 if change["new"] else 0.0)
                status.value = f"<span>Set {label} to {change['new']}</span>"
            except Exception as exc:
                status.value = f"<span style='color:#b00020'>Could not set {label}: {exc}</span>"

    checkbox.observe(_on_change, names="value")
    return checkbox


color_width, color_height = _parse_resolution(realsense_config.get("color_resolution", "640,480"), (640, 480))
fps = int(realsense_config.get("fps", "30"))

pipeline = rs2.pipeline()
config = rs2.config()
serial = realsense_config.get("serial", "")
if serial:
    config.enable_device(serial)
config.enable_stream(rs2.stream.color, color_width, color_height, rs2.format.rgb8, fps)

profile = pipeline.start(config)
device = profile.get_device()
color_sensor = _sensor_named(device, "rgb") or _sensor_named(device, "color")
depth_sensor = None
if color_sensor is None:
    color_sensor = _first_sensor_supporting(device, rs2.option.white_balance)
colorizer = rs2.colorizer()

image = widgets.Image(format="jpeg", width=color_width, height=color_height)
show_depth = widgets.Checkbox(value=False, description="Show depth", indent=False, disabled=True)
stop_button = widgets.Button(description="Stop RealSense live view", button_style="danger")
status = widgets.HTML(value="<span>RealSense live view running.</span>")

controls = []
for control in [
    _make_option_checkbox(color_sensor, rs2.option.enable_auto_exposure, "Auto exposure"),
    _make_option_slider(color_sensor, rs2.option.exposure, "Exposure", integer=True),
    _make_option_slider(color_sensor, rs2.option.gain, "Gain", integer=True),
    _make_option_checkbox(color_sensor, rs2.option.enable_auto_white_balance, "Auto white balance"),
    _make_option_slider(color_sensor, rs2.option.white_balance, "White balance", integer=True),
    _make_option_slider(depth_sensor, rs2.option.laser_power, "Laser power", integer=False),
    _make_option_slider(depth_sensor, rs2.option.visual_preset, "Depth preset", integer=True),
]:
    if control is not None:
        controls.append(control)

_rs_live = {"running": True, "pipeline": pipeline}


def stop_realsense_live(_=None):
    _rs_live["running"] = False
    try:
        pipeline.stop()
    except Exception:
        pass
    status.value = "<span>RealSense live view stopped.</span>"


stop_button.on_click(stop_realsense_live)


def _preview_loop():
    while _rs_live["running"]:
        try:
            ok, frames = pipeline.try_wait_for_frames(1000)
            if not ok:
                continue

            color_frame = frames.get_color_frame()
            if not color_frame:
                continue

            frame = np.asanyarray(color_frame.get_data())
            ok, encoded = cv2.imencode(".jpg", cv2.cvtColor(frame, cv2.COLOR_RGB2BGR))
            if ok:
                image.value = encoded.tobytes()
        except Exception as exc:
            if _rs_live["running"]:
                status.value = f"<span style='color:#b00020'>Preview stopped: {exc}</span>"
                stop_realsense_live()
            break
        time.sleep(0.01)


threading.Thread(target=_preview_loop, daemon=True).start()
display(widgets.VBox([image, widgets.HBox([stop_button, status]), widgets.VBox(controls)]))
```

### Calibrate RealSense color threshold
This uses the RealSense color stream in the same way the original `calibrate_camera.ipynb` used the embedded webcam: OpenCV trackbars tune a LAB color threshold, and the preview shows the selected pixels as white. Stop the live RealSense preview above before running these cells, because only one cell can own the camera stream at a time.

```python
import os
import time

os.environ.setdefault("QT_QPA_FONTDIR", "/usr/share/fonts/truetype/dejavu")

import cv2
import numpy as np

try:
    import pyrealsense2 as rs2
except ImportError as exc:
    raise ImportError("Install the Intel RealSense Python SDK first: %pip install pyrealsense2") from exc

_rs_color_pipeline = globals().get("_rs_color_pipeline", None)
_rs_color_format = globals().get("_rs_color_format", None)


def stop_realsense_streams(silent=False):
    """Stop RealSense streams started by this notebook kernel."""
    stopped_any = False

    if "_rs_live" in globals() and _rs_live.get("running"):
        _rs_live["running"] = False
        try:
            _rs_live["pipeline"].stop()
            stopped_any = True
        except Exception:
            pass
        try:
            status.value = "<span>RealSense live view stopped.</span>"
        except Exception:
            pass

    if "_rs_color_pipeline" in globals() and _rs_color_pipeline is not None:
        try:
            _rs_color_pipeline.stop()
            stopped_any = True
        except Exception:
            pass
        globals()["_rs_color_pipeline"] = None

    cv2.destroyAllWindows()
    for _ in range(4):
        cv2.waitKey(1)

    if stopped_any:
        time.sleep(0.8)
    if not silent:
        print("Stopped notebook-owned RealSense streams." if stopped_any else "No notebook-owned RealSense stream was running.")


def reset_realsense_device(wait_s=4.0):
    """Power-cycle RealSense devices through librealsense, then wait for reconnect."""
    stop_realsense_streams(silent=True)
    ctx = rs2.context()
    devices = list(ctx.query_devices())
    if not devices:
        print("No RealSense device found to reset.")
        return

    for device in devices:
        name = device.get_info(rs2.camera_info.name)
        serial = device.get_info(rs2.camera_info.serial_number)
        print(f"Resetting {name} ({serial})")
        device.hardware_reset()
    time.sleep(wait_s)
    print("Reset complete. Rerun the helper cell, then realsense_colorpicker().")


def _realsense_resolution(value, fallback):
    try:
        width, height = str(value).split(",")
        return int(width), int(height)
    except Exception:
        return fallback


def _realsense_color_settings():
    cfg = globals().get("realsense_config", {})
    width, height = _realsense_resolution(cfg.get("color_resolution", "640,480"), (640, 480))
    fps = int(cfg.get("fps", "30"))
    serial = cfg.get("serial", "")
    return serial, width, height, fps


def _candidate_color_modes(serial=None, width=None, height=None, fps=None):
    default_serial, default_width, default_height, default_fps = _realsense_color_settings()
    serial = default_serial if serial is None else serial
    width = default_width if width is None else width
    height = default_height if height is None else height
    fps = default_fps if fps is None else fps

    requested = [(width, height, fps)]
    fallbacks = [(640, 480, 15), (640, 480, 30), (424, 240, 15), (424, 240, 30), (320, 240, 15)]
    modes = []
    for mode in requested + fallbacks:
        if mode not in modes:
            modes.append(mode)

    formats = [(rs2.format.rgb8, "rgb8"), (rs2.format.bgr8, "bgr8")]
    for mode_width, mode_height, mode_fps in modes:
        for mode_format, format_name in formats:
            yield serial, mode_width, mode_height, mode_fps, mode_format, format_name


def _configure_color_stream(serial, width, height, fps, color_format):
    pipeline = rs2.pipeline()
    config = rs2.config()
    if serial:
        config.enable_device(serial)
    config.enable_stream(rs2.stream.color, width, height, color_format, fps)
    pipeline.start(config)
    return pipeline


def _read_realsense_bgr(pipeline, timeout_ms=1000):
    try:
        frames = pipeline.wait_for_frames(timeout_ms)
    except RuntimeError:
        return False, None

    color_frame = frames.get_color_frame()
    if not color_frame:
        return False, None

    frame = np.asanyarray(color_frame.get_data())
    if globals().get("_rs_color_format") == "bgr8":
        return True, frame
    return True, cv2.cvtColor(frame, cv2.COLOR_RGB2BGR)


def _read_realsense_bgr_with_retry(pipeline, total_timeout_s=8.0):
    deadline = time.time() + total_timeout_s
    while time.time() < deadline:
        ret, frame = _read_realsense_bgr(pipeline, timeout_ms=1000)
        if ret:
            return True, frame
        cv2.waitKey(1)
    return False, None


def _start_realsense_color_stream(serial=None, width=None, height=None, fps=None):
    stop_realsense_streams(silent=True)

    errors = []
    for mode_serial, mode_width, mode_height, mode_fps, mode_format, format_name in _candidate_color_modes(serial, width, height, fps):
        pipeline = None
        keep_pipeline = False
        try:
            pipeline = _configure_color_stream(mode_serial, mode_width, mode_height, mode_fps, mode_format)
            globals()["_rs_color_pipeline"] = pipeline
            globals()["_rs_color_format"] = format_name
            time.sleep(0.8)
            ret, _ = _read_realsense_bgr_with_retry(pipeline, total_timeout_s=5.0)
            if ret:
                keep_pipeline = True
                print(f"RealSense color stream: {mode_width}x{mode_height} @ {mode_fps} fps ({format_name})")
                return pipeline
            errors.append(f"{mode_width}x{mode_height}@{mode_fps} {format_name}: opened but no frames")
        except RuntimeError as exc:
            errors.append(f"{mode_width}x{mode_height}@{mode_fps} {format_name}: {exc}")
        finally:
            if pipeline is not None and not keep_pipeline:
                try:
                    pipeline.stop()
                except Exception:
                    pass
                globals()["_rs_color_pipeline"] = None
            if not keep_pipeline:
                time.sleep(0.3)

    raise RuntimeError("Could not start a RealSense color stream with frames. Tried:\n" + "\n".join(errors))


def _setup_lab_trackbars():
    cv2.namedWindow("Trackbars", 0)
    for limit_name, default_value in (("MIN", 0), ("MAX", 255)):
        for channel_name in "LAB":
            cv2.createTrackbar(f"{channel_name}_{limit_name}", "Trackbars", default_value, 255, lambda _value: None)


def _get_lab_trackbar_values():
    values = []
    for limit_name in ("MIN", "MAX"):
        for channel_name in "LAB":
            values.append(cv2.getTrackbarPos(f"{channel_name}_{limit_name}", "Trackbars"))
    return values


def _show_waiting_frame(window_name, width=640, height=160):
    frame = np.zeros((height, width, 3), dtype=np.uint8)
    cv2.putText(frame, "Waiting for RealSense color frames...", (20, 65), cv2.FONT_HERSHEY_SIMPLEX, 0.7, (255, 255, 255), 2)
    cv2.putText(frame, "Press ESC to stop", (20, 105), cv2.FONT_HERSHEY_SIMPLEX, 0.6, (180, 180, 180), 1)
    cv2.imshow(window_name, frame)


def realsense_colorpicker(serial=None, width=None, height=None, fps=None):
    try:
        pipeline = _start_realsense_color_stream(serial=serial, width=width, height=height, fps=fps)
    except RuntimeError as exc:
        raise RuntimeError(
            "Could not open a working RealSense color stream. Run reset_realsense_device(), "
            "restart the kernel if needed, and close other apps using the camera."
        ) from exc

    _setup_lab_trackbars()
    low_color = np.array([0, 0, 0], dtype=np.uint8)
    high_color = np.array([255, 255, 255], dtype=np.uint8)
    missed_since = None

    try:
        while True:
            ret, image = _read_realsense_bgr(pipeline, timeout_ms=1000)
            if not ret:
                if missed_since is None:
                    missed_since = time.time()
                    print("RealSense frame stream paused; waiting for frames again. Press ESC to stop.")
                _show_waiting_frame("Original")
                _show_waiting_frame("Thresh")
                if cv2.waitKey(100) & 0xFF == 27:
                    break
                continue

            if missed_since is not None:
                print(f"RealSense frames resumed after {time.time() - missed_since:.1f}s.")
                missed_since = None

            lab_frame = cv2.cvtColor(image, cv2.COLOR_BGR2LAB)
            l_min, a_min, b_min, l_max, a_max, b_max = _get_lab_trackbar_values()
            low_color = np.array([l_min, a_min, b_min], dtype=np.uint8)
            high_color = np.array([l_max, a_max, b_max], dtype=np.uint8)
            thresh = cv2.inRange(lab_frame, low_color, high_color)

            cv2.imshow("Original", cv2.flip(image, 1))
            cv2.imshow("Thresh", cv2.flip(thresh, 1))

            if cv2.waitKey(100) & 0xFF == 27:
                break
    finally:
        try:
            pipeline.stop()
        except Exception:
            pass
        globals()["_rs_color_pipeline"] = None
        cv2.destroyAllWindows()
        for _ in range(4):
            cv2.waitKey(1)

    return low_color, high_color


def locate_lab_color(img, low_color=None, high_color=None):
    if low_color is None:
        low_color = np.array([15, 55, 115], dtype=np.uint8)
    if high_color is None:
        high_color = np.array([240, 115, 205], dtype=np.uint8)

    lab_frame = cv2.cvtColor(img, cv2.COLOR_BGR2LAB)
    color_mask = cv2.inRange(lab_frame, low_color, high_color)
    contours, _ = cv2.findContours(color_mask, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

    if not contours:
        return None, None

    contour = max(contours, key=cv2.contourArea)
    x, y, w, h = cv2.boundingRect(contour)
    cv2.rectangle(img, (x, y), (x + w, y + h), (0, 0, 255), 2)
    cv2.circle(img, (round(x + w / 2), round(y + h / 2)), 5, (0, 0, 255), -1)
    return x, y


def show_realsense_color_detection(low_color, high_color, serial=None, width=None, height=None, fps=None):
    try:
        pipeline = _start_realsense_color_stream(serial=serial, width=width, height=height, fps=fps)
    except RuntimeError as exc:
        raise RuntimeError(
            "Could not open a working RealSense color stream. Run reset_realsense_device(), "
            "restart the kernel if needed, and close other apps using the camera."
        ) from exc

    cv2.namedWindow("RealSense color detection")
    missed_since = None

    try:
        while True:
            ret, frame = _read_realsense_bgr(pipeline, timeout_ms=1000)
            if not ret:
                if missed_since is None:
                    missed_since = time.time()
                    print("RealSense frame stream paused; waiting for frames again. Press ESC to stop.")
                _show_waiting_frame("RealSense color detection")
                if cv2.waitKey(100) & 0xFF == 27:
                    break
                continue

            if missed_since is not None:
                print(f"RealSense frames resumed after {time.time() - missed_since:.1f}s.")
                missed_since = None

            locate_lab_color(frame, low_color, high_color)
            cv2.imshow("RealSense color detection", frame)

            if cv2.waitKey(100) & 0xFF == 27:
                break
    finally:
        try:
            pipeline.stop()
        except Exception:
            pass
        globals()["_rs_color_pipeline"] = None
        cv2.destroyAllWindows()
        for _ in range(4):
            cv2.waitKey(1)


def realsense_camera_tuner(serial=None, width=None, height=None, fps=None):
    """OpenCV-only RealSense preview with keyboard camera-parameter controls."""
    try:
        pipeline = _start_realsense_color_stream(serial=serial, width=width, height=height, fps=fps)
    except RuntimeError as exc:
        raise RuntimeError(
            "Could not open a working RealSense color stream. Run reset_realsense_device(), "
            "restart the kernel if needed, and close other apps using the camera."
        ) from exc

    profile = pipeline.get_active_profile()
    device = profile.get_device()
    color_sensor = None
    for sensor in device.query_sensors():
        if sensor.supports(rs2.option.exposure) or sensor.supports(rs2.option.gain):
            color_sensor = sensor
            break

    manual_exposure = False
    exposure = 100.0
    gain = 64.0
    white_balance = 4600.0

    def set_if_supported(option, value):
        if color_sensor is not None and color_sensor.supports(option):
            opt_range = color_sensor.get_option_range(option)
            value = min(max(float(value), opt_range.min), opt_range.max)
            color_sensor.set_option(option, value)
            return value
        return value

    try:
        if color_sensor is not None and color_sensor.supports(rs2.option.enable_auto_exposure):
            manual_exposure = not bool(color_sensor.get_option(rs2.option.enable_auto_exposure))
        if color_sensor is not None and color_sensor.supports(rs2.option.exposure):
            exposure = color_sensor.get_option(rs2.option.exposure)
        if color_sensor is not None and color_sensor.supports(rs2.option.gain):
            gain = color_sensor.get_option(rs2.option.gain)
        if color_sensor is not None and color_sensor.supports(rs2.option.white_balance):
            white_balance = color_sensor.get_option(rs2.option.white_balance)
    except Exception:
        pass

    cv2.namedWindow("RealSense camera tuner")
    print("Controls: e=auto/manual exposure, [/] exposure, -/= gain, ,/. white balance, ESC=stop")

    try:
        while True:
            ret, frame = _read_realsense_bgr(pipeline, timeout_ms=1000)
            if ret:
                overlay = [
                    "e auto/manual exposure | [ ] exposure | - = gain | , . white balance | ESC stop",
                    f"manual_exposure={manual_exposure} exposure={exposure:.0f} gain={gain:.0f} white_balance={white_balance:.0f}",
                ]
                for i, text in enumerate(overlay):
                    y = 26 + i * 26
                    cv2.putText(frame, text, (10, y), cv2.FONT_HERSHEY_SIMPLEX, 0.55, (0, 0, 0), 3)
                    cv2.putText(frame, text, (10, y), cv2.FONT_HERSHEY_SIMPLEX, 0.55, (255, 255, 255), 1)
                cv2.imshow("RealSense camera tuner", frame)
            else:
                _show_waiting_frame("RealSense camera tuner")

            key = cv2.waitKey(100) & 0xFF
            if key == 27:
                break
            if color_sensor is None:
                continue
            if key == ord("e") and color_sensor.supports(rs2.option.enable_auto_exposure):
                manual_exposure = not manual_exposure
                color_sensor.set_option(rs2.option.enable_auto_exposure, 0.0 if manual_exposure else 1.0)
            elif key == ord("["):
                manual_exposure = True
                if color_sensor.supports(rs2.option.enable_auto_exposure):
                    color_sensor.set_option(rs2.option.enable_auto_exposure, 0.0)
                exposure = set_if_supported(rs2.option.exposure, exposure - 10)
            elif key == ord("]"):
                manual_exposure = True
                if color_sensor.supports(rs2.option.enable_auto_exposure):
                    color_sensor.set_option(rs2.option.enable_auto_exposure, 0.0)
                exposure = set_if_supported(rs2.option.exposure, exposure + 10)
            elif key == ord("-"):
                gain = set_if_supported(rs2.option.gain, gain - 4)
            elif key in (ord("="), ord("+")):
                gain = set_if_supported(rs2.option.gain, gain + 4)
            elif key == ord(","):
                white_balance = set_if_supported(rs2.option.white_balance, white_balance - 100)
            elif key == ord("."):
                white_balance = set_if_supported(rs2.option.white_balance, white_balance + 100)
    finally:
        try:
            pipeline.stop()
        except Exception:
            pass
        globals()["_rs_color_pipeline"] = None
        cv2.destroyAllWindows()
        for _ in range(4):
            cv2.waitKey(1)


print("RealSense color calibration helpers loaded. Run realsense_colorpicker(). If the camera is stuck, run reset_realsense_device() first.")
```

### Discover modules
Next, we will print a list of discovered modules:

```python
moduleids = api.discoverModules()
print("Module IDs: ", moduleids)

if not moduleids:
    raise RuntimeError("No Fable modules found. Check that the module is powered on and connected to the dongle.")
```

### Move
We can move the joints with the `setPos(posX, posY, moduleID)` command.

The arguments are:

`posX`: X motor angle (degrees)

`posY`: Y motor angle (degrees)

`moduleID`: The module ID. You can use `moduleids[0]` to use the first discovered module.

```python
moduleID = moduleids[0]

speed = 50  # 1-100 percent; 100 is fastest
api.setSpeed(speed, speed, moduleID)
api.setPos(0, 70, moduleID)
```

```python
moduleID = moduleids[0]

speed = 100  # 1-100 percent; 100 is fastest
api.setSpeed(speed, speed, moduleID)
api.setPos(0, 15, moduleID)
```

### Changing the reference
We can try to make the joints follow a changing reference angle, using a for loop:

```python
for i in range(10):
    api.setPos(i*10, i*5, moduleID)
```

What happened? All of the commands were sent immediately, giving the module no time to react.

To not send all commands at once, we can wait in between commands. How much time is required? Try different sleep durations


```python
for i in range(-9,9,1):
    api.setPos(i*10, i*7, moduleID)
    api.sleep(2)
```
